```python
import numpy as np
#from sympy import legendre_symbol     #If the two lines are commented out, s has to be a quadratic nonresidue

def M(p,s):  #p is a prime and s is a quandratic non-residue mod p
    w = np.exp(2j * np.pi / p)
    M = np.matrix(np.zeros((2*p,2*p), dtype=complex))
    for i in range(p):
        for j in range(p):
            M[  i,   j] = w ** ((i * j)%p)
            M[  i, p+j] = w ** ((i**2*(s-1)+i*j)%p)
            M[p+i,   j] = w ** ((-2**(p-3)*(j-i)**2)%p)
            M[p+i, p+j] = w ** ((-2**(p-3)*s**(p-2)*(j-i)**2)%p)
#            M[p+i, p+j] = - legendre_symbol(s,p) * w ** ((-2**(p-3)*s**(p-2)*(j-i)**2)%p)
    return(M)
```

```python
def diagram(H):
    HH=np.array(H.H)
    HT=np.array(H.T)
    H=np.array(H)
    A=np.sum(HH[:,:,None,None,None]**2*
             H[None,:,:,None,None]*
             HH[None,None,:,:,None]**2*
             HT[:,None,None,:,None]*
             H[None,None,None,:,:]
                  ,axis=(3))
    A=np.sum(A[:,:,:,:,None,None]*
             HH[None,None,None,:,:,None]**2*
              H[None,None,None,None,:,:]*
             HT[None,None,:,None,:,None]
             ,axis=(4))
    return np.sum(A[:,:,:,:,:,None]*
                  HH[None,None,None,None,:,:]**2*
                  H[None,:,None,None,:,None]*
                  HT[:,None,None,None,None,:]*
                  HT[None,None,None,:,None,:]
                 ,axis=(0,1,2,3,4,5))
```
